```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import log_loss,r2_score,mean_squared_error
from sklearn.linear_model import LinearRegression
```

```python
df=pd.read_csv("/Users/gurumithra.s.s/Desktop/supervisedlearning/csv files/it_company.csv")
df
```

```python
lb=LabelEncoder()
df['Education_Level']=lb.fit_transform(df['Education_Level'])
df['Department']=lb.fit_transform(df['Department'])
df['Gender']=lb.fit_transform(df['Gender'])
```

```python
x=df.drop("Salary",axis=1)
y=df["Salary"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
y_pred
```

```python
score=r2_score(y_pred,y_test)
mse=mean_squared_error(y_pred,y_test)
rmse=np.sqrt(mse)
print("R2 Score:",score)
print("Mean Squared Error:",mse)
print("Root Mean Squared Error:",rmse)
```

```python
plt.scatter(y_test,y_pred)
plt.plot([y_test.min(),y_test.max()],[y_pred.min(),y_pred.max()],color='red')
plt.show()
```
